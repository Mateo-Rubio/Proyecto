```python
import pandas as pd
import matplotlib.pyplot as plt
```

### Downloading the 794nm-14.8mW-FFT.csv file

```python
url_794nm = 'https://raw.githubusercontent.com/Mateo-Rubio/Proyecto/refs/heads/master/LaserFFT/data/794nm-14.8mW-FFT.csv'
df_794nm = pd.read_csv(url_794nm, names = ["Frequency[Hz]", "Valor"], skiprows=21 )
display(df_794nm.head())
```

### Downloading the 852nm-19.5mW-FFT.csv file

```python
url_852nm = 'https://raw.githubusercontent.com/Mateo-Rubio/Proyecto/refs/heads/master/LaserFFT/data/852nm-19.5mW-FFT.csv'
df_852nm = pd.read_csv(url_852nm, names = ["Frequency[Hz]", "Valor"], skiprows=21 )
display(df_852nm.head())
```

```python
plt.plot(df_852nm["Frequency[Hz]"]*1e-9, df_852nm["Valor"])
plt.title(r"FFT de láser con $\lambda = 852$nm ")
plt.xlabel("Frecuencias del láser [GHz]")
plt.ylabel("Magnitud de las frecuencias[dBV RMS]")
plt.show()
```

```python
plt.plot(df_794nm["Frequency[Hz]"]*1e-9, df_794nm["Valor"])
plt.title(r"FFT de láser con $\lambda = 794$nm ")
plt.xlabel("Frecuencias del láser [GHz]")
plt.ylabel("Magnitud de las frecuencias[dBV RMS]")
plt.show()
```
